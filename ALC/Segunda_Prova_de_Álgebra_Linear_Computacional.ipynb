```python
import pandas as pd  # Biblioteca para manipulação de dados
import numpy as np  # Biblioteca para computação numérica
from google.colab import files  # Módulo para upload de arquivos no Google Colab

def cholesky(A):
    n = A.shape[0]  # Número de linhas/colunas da matriz A
    L = np.zeros((n, n))  # Inicializa a matriz L com zeros

    # Verificar se A é simétrica
    if not np.allclose(A, A.T):  # Verifica se A é aproximadamente igual à sua transposta
        print("A matriz fornecida não é simétrica.")  # Mensagem de erro
        return None  # Retorna None

    # Verificar se A é definida positiva
    for i in range(n):
        sum = 0.0  # Inicializa a soma dos quadrados dos elementos da linha i
        for k in range(i):
            sum += L[i][k] ** 2  # Adiciona o quadrado de L[i][k] à soma
        diagonal = A[i, i] - sum  # Calcula o valor da diagonal de L
        if diagonal <= 0:  # Verifica se a diagonal é menor ou igual a zero
            print("A matriz fornecida não é definida positiva.")  # Mensagem de erro
            return None  # Retorna None
        L[i, i] = np.sqrt(diagonal)  # Define o elemento diagonal de L
        for j in range(i + 1, n):
            sum = 0.0  # Inicializa a soma dos produtos
            for k in range(i):
                sum += L[i, k] * L[j, k]  # Adiciona o produto L[i, k] * L[j, k] à soma
            if L[i, i] == 0:  # Verifica se o elemento diagonal de L é zero
                print("A matriz fornecida não é definida positiva.")  # Mensagem de erro
                return None  # Retorna None
            L[j, i] = (A[j, i] - sum) / L[i, i]  # Calcula o elemento fora da diagonal de L

    return L  # Retorna a matriz L


if __name__ == "__main__":
  # Carregar o arquivo Excel
  print("Faça upload da matriz A")
  uploaded = files.upload()  # Pede ao usuário para fazer upload do arquivo

  # Obter o nome do arquivo
  file_name = list(uploaded.keys())[0]  # Obtém o nome do arquivo carregado

  # Ler a matriz A do arquivo Excel
  df_A = pd.read_excel(file_name, header=None)  # Lê o arquivo Excel e armazena em um DataFrame

  # Converter o dataframe para um array numpy
  A = df_A.to_numpy()  # Converte o DataFrame para um array numpy
  print("Matriz original:")  # Imprime o texto "Matriz original:"
  print(np.matrix(A))  # Imprime a matriz A

  L = cholesky(A)  # Chama a função cholesky para calcular o fator de Cholesky de A

  if L is not None:  # Verifica se a decomposição foi bem-sucedida
    print("\nMatriz L (fator de Cholesky):")  # Imprime o texto "Matriz L (fator de Cholesky):"
    print(np.matrix(L))  # Imprime a matriz L

    print("\n\nMultiplicando o L com a sua transposta")
    print(np.dot(L, L.T))
```

Letra C (Fatoração de LU)

```python
import numpy as np  # Biblioteca para computação numérica
import pandas as pd  # Biblioteca para manipulação de dados
from google.colab import files  # Módulo para upload de arquivos no Google Colab

def determinante(A):
    return np.linalg.det(A)  # Retorna o determinante da matriz A usando a função det da NumPy

def decomposicao_LU(A):
    n = A.shape[0]  # Número de linhas da matriz A
    L = np.zeros((n, n))  # Inicializa a matriz L com zeros
    U = np.zeros((n, n))  # Inicializa a matriz U com zeros

    # Verificar se A é quadrada
    if A.shape[0] != A.shape[1]:  # Verifica se o número de linhas é diferente do número de colunas
        print("A matriz fornecida não é quadrada.")  # Mensagem de erro
        return None, None  # Retorna None para L e U

    # Verificar se A é não singular
    if determinante(A) == 0:  # Verifica se o determinante de A é zero
        print("A matriz fornecida é singular (determinante igual a zero).")  # Mensagem de erro
        return None, None  # Retorna None para L e U

    # Fatoração LU
    for i in range(n):
        # Matriz U
        for j in range(i, n):
            sum = 0  # Inicializa a soma
            for k in range(i):
                sum += L[i, k] * U[k, j]  # Calcula a soma
            U[i, j] = A[i, j] - sum  # Define os elementos de U

        # Matriz L
        L[i, i] = 1  # Define a diagonal principal de L como 1
        for j in range(i + 1, n):
            sum = 0  # Inicializa a soma
            for k in range(i):
                sum += L[j, k] * U[k, i]  # Calcula a soma
            L[j, i] = (A[j, i] - sum) / U[i, i]  # Define os elementos de L

    return L, U  # Retorna as matrizes L e U



if __name__ == "__main__":
  # Carregar o arquivo Excel
  print("Faça upload da matriz A")
  uploaded = files.upload()  # Pede ao usuário para fazer upload do arquivo

  # Obter o nome do arquivo
  file_name = list(uploaded.keys())[0]  # Obtém o nome do arquivo carregado

  # Ler a matriz A do arquivo Excel
  df_A = pd.read_excel(file_name, header=None)  # Lê o arquivo Excel e armazena em um DataFrame

  # Converter o dataframe para um array numpy
  A = df_A.to_numpy()  # Converte o DataFrame para um array numpy

  print("Matriz original:")  # Imprime o texto "Matriz original:"
  print(A)  # Imprime a matriz A

  L, U = decomposicao_LU(A)  # Chama a função decomposicao_LU para calcular a fatoração LU de A

  if L is not None and U is not None:  # Verifica se a decomposição foi bem-sucedida
    print("\nMatriz L (fator de LU):")  # Imprime o texto "Matriz L (fator de LU):"
    print(L)  # Imprime a matriz L
    print("\nMatriz U (fator de LU):")  # Imprime o texto "Matriz U (fator de LU):"
    print(U)  # Imprime a matriz U

    print("\n\nL * U =")
    print(np.dot(L, U))
```

Letra D (Métodos Gauss-Seidel/Sassenfeld)

```python
import numpy as np  # Importa a biblioteca NumPy para manipulação de arrays
import pandas as pd  # Importa a biblioteca Pandas para leitura de dados tabulares
from google.colab import files  # Importa o módulo files do Google Colab para upload de arquivos

# Função para verificar o critério de Sassenfeld
def criterio_sassenfeld(A):
    n = A.shape[0]  # Obtém o número de linhas da matriz A
    beta = np.zeros(n)  # Inicializa um vetor para armazenar os valores de beta
    max_beta = 0  # Inicializa o valor máximo de beta

    # Loop para calcular os valores de beta para cada linha de A
    for i in range(n):
        soma = 0
        # Calcula a soma dos valores absolutos dos elementos fora da diagonal
        for j in range(n):
            if j != i:
                soma += abs(A[i, j])
        # Calcula beta para a linha i
        beta[i] = soma / abs(A[i, i])
        # Atualiza o valor máximo de beta
        if beta[i] > max_beta:
            max_beta = beta[i]

    # Retorna True se o maior valor de beta for menor que 1, indicando que o critério de Sassenfeld é satisfeito
    return max_beta < 1

# Função para verificar se a matriz é positiva definida
def is_positive_definite(A):
    # Verifica se todos os autovalores de A são maiores que zero
    return np.all(np.linalg.eigvals(A) > 0)

# Função para resolver um sistema linear usando o método de Gauss-Seidel
def gauss_seidel(A, b, x0=None, tol=1e-6, max_iter=1000):
    n = len(b)  # Obtém o tamanho do vetor b, que é o número de linhas da matriz A
    if x0 is None:
        x0 = np.zeros(n)  # Inicializa o vetor de solução inicial x0 com zeros

    x = np.copy(x0)  # Faz uma cópia do vetor de solução inicial para x
    x_novo = np.zeros(n)  # Inicializa um vetor para armazenar a solução atualizada x

    # Loop principal do método de Gauss-Seidel
    for iteracao in range(1, max_iter + 1):  # Executa um número máximo de iterações
        # Loop para calcular os novos valores de x
        for i in range(n):
            # Calcula o novo valor de x[i]
            x_novo[i] = (b[i] - np.dot(A[i, :i], x_novo[:i]) - np.dot(A[i, i + 1:], x[i + 1:])) / A[i, i]

        # Verifica a convergência comparando o erro entre x_novo e x com a tolerância especificada
        if np.linalg.norm(x_novo - x) < tol:
            print(f"Convergiu após {iteracao} iterações.")
            return x_novo  # Retorna a solução se o erro for menor que a tolerância

        # Atualiza x com os novos valores calculados
        x = np.copy(x_novo)

    # Se o método não convergir dentro do número máximo de iterações especificado, levanta uma exceção
    raise ValueError("O método de Gauss-Seidel não convergiu dentro da tolerância especificada.")

# Código principal
if __name__ == "__main__":
    print("Faça upload da matriz A.")
    uploaded_A = files.upload()  # Solicita ao usuário para fazer upload do arquivo da matriz A

    print("Faça upload do vetor b.")
    uploaded_b = files.upload()  # Solicita ao usuário para fazer upload do arquivo do vetor b

    # Obtém o nome do arquivo da matriz A e do vetor b
    nome_arquivo_A = list(uploaded_A.keys())[0]
    nome_arquivo_b = list(uploaded_b.keys())[0]

    # Lê os arquivos Excel da matriz A e do vetor b
    df_A = pd.read_excel(nome_arquivo_A, header=None)
    df_b = pd.read_excel(nome_arquivo_b, header=None)

    # Converte os DataFrames da matriz A e do vetor b para arrays NumPy
    A = df_A.to_numpy()
    b = df_b.to_numpy().flatten()

    # Verifica se a matriz A é positiva definida e se satisfaz o critério de Sassenfeld
    if not is_positive_definite(A):
        print("A matriz A não é positiva definida.")
    elif not criterio_sassenfeld(A):
        print("A matriz A não satisfaz o critério de Sassenfeld. O método pode não convergir.")
    else:
        # Parâmetros do método de Gauss-Seidel
        tolerancia = 1e-6
        max_iteracoes = 1000

        # Resolve o sistema linear usando o método de Gauss-Seidel
        solucao = gauss_seidel(A, b, tol=tolerancia, max_iter=max_iteracoes)

        # Imprime a solução encontrada
        print("Solução:")
        for i, valor in enumerate(solucao):
            print(f"x[{i}] = {valor}")
```

Letra G (Número condição da matriz e uma estimativa)

```python
import numpy as np  # Biblioteca para computação numérica
from google.colab import files  # Módulo para upload de arquivos no Google Colab
import pandas as pd  # Biblioteca para manipulação de dados

# Classe Matriz para representar e operar com matrizes
class Matriz:
    def __init__(self, dados):
        # Inicializa a matriz com os dados fornecidos
        self.dados = np.array(dados)
        self.linhas, self.colunas = self.dados.shape

    def transposta(self):
        # Retorna a transposta da matriz
        return Matriz(self.dados.T)

    def svd(self):
        # Calcula a decomposição em valores singulares (SVD) da matriz usando a biblioteca NumPy
        U, s, Vt = np.linalg.svd(self.dados)
        # Converte s para uma matriz diagonal
        S = np.zeros((self.linhas, self.colunas))
        S[:min(self.linhas, self.colunas), :min(self.linhas, self.colunas)] = np.diag(s)
        # Retorna as matrizes U, S (valores singulares) e Vt (transposta de V)
        return Matriz(U), Matriz(S), Matriz(Vt)

def calcular_numero_condicao(matriz):
    # Calcula o número de condição da matriz usando a decomposição em valores singulares (SVD)
    _, S, _ = matriz.svd()
    # Obtém os valores singulares
    valores_singulares = np.diag(S.dados)
    # Verifica se a matriz é singular (possui valor singular zero)
    if np.any(valores_singulares == 0):
        return "A matriz é singular."
    # Calcula o número de condição
    maior_valor_singular = valores_singulares[0]
    menor_valor_singular = valores_singulares[-1]
    numero_condicao = maior_valor_singular / menor_valor_singular
    return numero_condicao

if __name__ == "__main__":
    # Carregar arquivos Excel
    print("Faça upload da matriz A.")
    uploaded_A = files.upload()

    # Ler a matriz A do arquivo Excel
    nome_arquivo_A = list(uploaded_A.keys())[0]
    df_A = pd.read_excel(nome_arquivo_A, header=None)

    # Converter os dados do DataFrame para um array numpy
    dados = df_A.to_numpy()

    # Criando uma instância da classe Matriz
    matriz = Matriz(dados)

    # Verifica se a matriz é quadrada
    if matriz.linhas != matriz.colunas:
        print("A matriz não é quadrada.")
    else:
        # Calculando o número de condição da matriz
        resultado = calcular_numero_condicao(matriz)

        # Exibindo o resultado
        print("Resultado:", resultado)

```

Letra H (Produto interno e distância entre esses dois vetores)

```python
import math  # Importa a biblioteca math para operações matemáticas
from google.colab import files  # Importa o módulo files do Google Colab para upload de arquivos
import pandas as pd  # Importa a biblioteca pandas para manipulação de dados

def verificar_tamanho_vetores(vetor1, vetor2):
    """
    Verifica se os vetores têm o mesmo tamanho.

    Parâmetros:
    vetor1 (list): Primeiro vetor.
    vetor2 (list): Segundo vetor.

    Levanta:
    ValueError: Se os vetores não tiverem o mesmo tamanho.
    """
    if len(vetor1) != len(vetor2):  # Verifica se os vetores têm o mesmo tamanho
        raise ValueError("Os vetores devem ter o mesmo tamanho")  # Lança um erro se os vetores não tiverem o mesmo tamanho

def produto_interno(vetor1, vetor2):
    """
    Calcula o produto interno entre dois vetores.

    Parâmetros:
    vetor1 (list): Primeiro vetor.
    vetor2 (list): Segundo vetor.

    Retorna:
    float: O produto interno dos vetores.
    """
    verificar_tamanho_vetores(vetor1, vetor2)  # Verifica se os vetores têm o mesmo tamanho

    produto = 0  # Inicializa o produto interno como zero
    for i in range(len(vetor1)):  # Itera sobre os elementos dos vetores
        produto += vetor1[i] * vetor2[i]  # Calcula o produto interno somando os produtos dos elementos correspondentes
    return produto  # Retorna o produto interno

def distancia_entre_vetores(vetor1, vetor2):
    """
    Calcula a distância Euclidiana entre dois vetores.

    Parâmetros:
    vetor1 (list): Primeiro vetor.
    vetor2 (list): Segundo vetor.

    Retorna:
    float: A distância Euclidiana entre os vetores.
    """
    verificar_tamanho_vetores(vetor1, vetor2)  # Verifica se os vetores têm o mesmo tamanho

    soma_dos_quadrados = 0  # Inicializa a soma dos quadrados das diferenças como zero
    for i in range(len(vetor1)):  # Itera sobre os elementos dos vetores
        soma_dos_quadrados += (vetor1[i] - vetor2[i]) ** 2  # Calcula a soma dos quadrados das diferenças dos elementos correspondentes
    return math.sqrt(soma_dos_quadrados)  # Retorna a raiz quadrada da soma dos quadrados das diferenças, ou seja, a distância Euclidiana

if __name__ == "__main__":
    # Carregar arquivos Excel
    print("Faça upload do Vetor b.")  # Solicita o upload do vetor b
    uploaded_a = files.upload()  # Realiza o upload do arquivo
    print("Faça upload do Vetor c.")  # Solicita o upload do vetor c
    uploaded_b = files.upload()  # Realiza o upload do arquivo

    # Ler o vetor b e o vetor c dos arquivos Excel
    nome_arquivo_b = list(uploaded_a.keys())[0]  # Obtém o nome do arquivo do vetor b
    nome_arquivo_c = list(uploaded_b.keys())[0]  # Obtém o nome do arquivo do vetor c

    df_b = pd.read_excel(nome_arquivo_b, header=None)  # Lê o vetor b do arquivo Excel
    df_c = pd.read_excel(nome_arquivo_c, header=None)  # Lê o vetor c do arquivo Excel

    # Converter os DataFrames para arrays numpy e achatar (flatten) para obter vetores
    vetorB = df_b.to_numpy().flatten()  # Converte o DataFrame do vetor b para um array numpy e achata para obter um vetor
    vetorC = df_c.to_numpy().flatten()  # Converte o DataFrame do vetor c para um array numpy e achata para obter um vetor

    # Calculando o produto interno
    produto = produto_interno(vetorB, vetorC)  # Calcula o produto interno dos vetores b e c
    print("Produto Interno de <b, c>: ", produto)  # Imprime o produto interno

    # Calculando a distância entre os vetores
    distancia = distancia_entre_vetores(vetorB, vetorC)  # Calcula a distância Euclidiana entre os vetores b e c
    print("Distância entre os vetores: ", distancia)  # Imprime a distância entre os vetores
```

Letra I (Distância entre A e P)

```python
import math  # Importa a biblioteca math para operações matemáticas
from google.colab import files  # Importa o módulo files do Google Colab para upload de arquivos
import pandas as pd  # Importa a biblioteca pandas para manipulação de dados

def distancia_entre_matrizes(matrizA, matrizP):
    """
    Calcula a distância Euclidiana entre duas matrizes.

    Parâmetros:
    matrizA (list of list of float): Primeira matriz.
    matrizP (list of list of float): Segunda matriz.

    Retorna:
    float: A distância Euclidiana entre as matrizes.

    Levanta:
    ValueError: Se as matrizes não tiverem as mesmas dimensões.
    """
    if len(matrizA) != len(matrizP) or len(matrizA[0]) != len(matrizP[0]):  # Verifica se as matrizes têm as mesmas dimensões
        raise ValueError("As matrizes devem ter as mesmas dimensões")  # Lança um erro se as matrizes não tiverem as mesmas dimensões

    soma_das_distancias = 0  # Inicializa a soma das distâncias como zero
    for i in range(len(matrizA)):  # Itera sobre as linhas das matrizes
        for j in range(len(matrizA[0])):  # Itera sobre as colunas das matrizes
            soma_das_distancias += (matrizA[i][j] - matrizP[i][j]) ** 2  # Calcula a soma dos quadrados das diferenças dos elementos correspondentes
    return math.sqrt(soma_das_distancias)  # Retorna a raiz quadrada da soma das distâncias, ou seja, a distância Euclidiana

if __name__ == "__main__":
    # Carregar arquivos Excel
    print("Faça upload da matriz A.")  # Solicita o upload da matriz A
    uploaded_a = files.upload()  # Realiza o upload do arquivo
    print("Faça upload da matriz P.")  # Solicita o upload da matriz P
    uploaded_p = files.upload()  # Realiza o upload do arquivo

    # Ler a matriz A e a matriz P dos arquivos Excel
    nome_arquivo_a = list(uploaded_a.keys())[0]  # Obtém o nome do arquivo da matriz A
    nome_arquivo_p = list(uploaded_p.keys())[0]  # Obtém o nome do arquivo da matriz P

    df_a = pd.read_excel(nome_arquivo_a, header=None)  # Lê a matriz A do arquivo Excel
    df_p = pd.read_excel(nome_arquivo_p, header=None)  # Lê a matriz P do arquivo Excel

    # Converter os DataFrames para arrays numpy
    matrizA = df_a.to_numpy()  # Converte o DataFrame da matriz A para um array numpy
    matrizP = df_p.to_numpy()  # Converte o DataFrame da matriz P para um array numpy

    # Calculando a distância entre as matrizes
    distancia = distancia_entre_matrizes(matrizA, matrizP)  # Calcula a distância Euclidiana entre as matrizes A e P
    print("Distância entre as matrizes: ", distancia)  # Imprime a distância entre as matrizes
```

Letra J(Decomposição QR)

```python
import numpy as np  # Importa a biblioteca NumPy para operações numéricas
import pandas as pd  # Importa a biblioteca pandas para manipulação de dados
from google.colab import files  # Importa o módulo files do Google Colab para upload de arquivos

def verificar_condicoes(A):
    """
    Verifica se a matriz A atende às condições necessárias para a decomposição QR.

    Parâmetros:
    A (numpy.ndarray): Matriz a ser verificada.

    Retorna:
    bool: True se a matriz atender às condições, False caso contrário.
    """
    m, n = A.shape  # Obtém as dimensões da matriz A

    # Condição 1: A matriz deve ser quadrada ou retangular (m >= n)
    if m < n:  # Verifica se o número de linhas é menor que o número de colunas
        print("Erro: A matriz não é retangular (número de linhas < número de colunas).")
        return False  # Retorna False se a condição não for atendida

    # Condição 2: A matriz deve ter posto completo
    for i in range(n):
        if np.all(A[:, i] == 0):  # Verifica se alguma coluna é completamente zero
            print("Erro: A matriz não tem posto completo.")
            return False  # Retorna False se a condição não for atendida

    return True  # Retorna True se todas as condições forem atendidas

def decomposicao_QR(A):
    """
    Realiza a decomposição QR da matriz A.

    Parâmetros:
    A (numpy.ndarray): Matriz a ser decomposta.

    Retorna:
    tuple: Matrizes Q e R resultantes da decomposição QR.
    """
    m, n = A.shape  # Obtém as dimensões da matriz A

    Q = np.zeros((m, n))  # Inicializa a matriz Q com zeros
    R = np.zeros((n, n))  # Inicializa a matriz R com zeros

    # Implementação da decomposição QR (algoritmo real)
    for j in range(n):  # Itera sobre as colunas de A
        coluna = A[:, j].copy()  # Copia a j-ésima coluna de A
        for k in range(j):  # Itera sobre as colunas anteriores
            produto_escalar = np.dot(Q[:, k], coluna)  # Calcula o produto escalar entre Q[:, k] e a coluna
            coluna = coluna - produto_escalar * Q[:, k]  # Atualiza a coluna
        norma_coluna = np.linalg.norm(coluna)  # Calcula a norma da coluna
        Q[:, j] = coluna / norma_coluna  # Normaliza a coluna e armazena em Q
        for i in range(j + 1):  # Atualiza a matriz R
            R[i, j] = np.dot(Q[:, i], A[:, j])  # Calcula o elemento R[i, j]

    return Q, R  # Retorna as matrizes Q e R

if __name__ == "__main__":
    # Carregar o arquivo Excel
    print("Faça o upload da matriz A")  # Pede ao usuário para fazer o upload da matriz A
    uploaded = files.upload()  # Realiza o upload do arquivo

    # Obter o nome do arquivo
    file_name = list(uploaded.keys())[0]  # Obtém o nome do arquivo carregado

    # Ler a matriz A do arquivo Excel
    df_A = pd.read_excel(file_name, header=None)  # Lê o arquivo Excel em um DataFrame

    # Converter o dataframe para um array numpy
    A = df_A.to_numpy()  # Converte o DataFrame para um array numpy

    if not verificar_condicoes(A):  # Verifica se a matriz atende às condições necessárias
        print("Erro: A matriz fornecida não atende às condições para a decomposição QR.")  # Exibe uma mensagem de erro
    else:
        Q, R = decomposicao_QR(A)  # Realiza a decomposição QR

        # Imprimir as matrizes Q e R
        print("Matriz Q:")  # Imprime uma mensagem
        print(Q)  # Imprime a matriz Q
        print("Matriz R:")  # Imprime uma mensagem
        print(R)  # Imprime a matriz R
        print(np.dot(Q,R))
```

Letra K (Critério das linhas, das colunas e de Sassenfeld)

```python
import pandas as pd  # Importa a biblioteca pandas para manipulação de dados
from google.colab import files  # Importa o módulo files do Google Colab para upload de arquivos

def criterio_linhas(matriz):
    """
    Verifica a convergência de uma matriz usando o critério das linhas.

    Parâmetros:
    matriz (list of list of float): Matriz dos coeficientes.

    Retorna:
    bool: True se a convergência for garantida, False caso contrário.
    """
    num_linhas = len(matriz)  # Obtém o número de linhas da matriz
    num_colunas = len(matriz[0])  # Obtém o número de colunas da matriz

    # Itera sobre as linhas
    for i in range(num_linhas):
        elemento_diagonal = abs(matriz[i][i])  # Elemento da diagonal principal
        soma_outros = 0

        # Soma os outros elementos da linha
        for j in range(num_colunas):
            if j != i:  # Ignora o elemento diagonal
                soma_outros += abs(matriz[i][j])

        # Verifica se o critério é atendido
        if elemento_diagonal <= soma_outros:
            return False  # Não converge

    return True  # Converge


def criterio_colunas(matriz):
    """
    Verifica a convergência de uma matriz usando o critério das colunas.

    Parâmetros:
    matriz (list of list of float): Matriz dos coeficientes.

    Retorna:
    bool: True se a convergência for garantida, False caso contrário.
    """
    num_linhas = len(matriz)  # Obtém o número de linhas da matriz
    num_colunas = len(matriz[0])  # Obtém o número de colunas da matriz

    # Itera sobre as colunas
    for j in range(num_colunas):
        elemento_diagonal = abs(matriz[j][j])  # Elemento da diagonal principal
        soma_outros = 0

        # Soma os outros elementos da coluna
        for i in range(num_linhas):
            if i != j:  # Ignora o elemento diagonal
                soma_outros += abs(matriz[i][j])

        # Verifica se o critério é atendido
        if elemento_diagonal <= soma_outros:
            return False  # Não converge

    return True  # Converge


class ResultadoSassenfeld:
    """
    Classe para armazenar o resultado do critério de Sassenfeld.
    """
    def __init__(self, criterio_aplicavel, criterio_atendido, max_beta):
        self.criterio_aplicavel = criterio_aplicavel  # Se o critério é aplicável
        self.criterio_atendido = criterio_atendido  # Se o critério é atendido
        self.max_beta = max_beta  # O valor máximo de beta


def criterio_sassenfeld(matriz_coeficientes):
    """
    Verifica a convergência de uma matriz usando o critério de Sassenfeld.

    Parâmetros:
    matriz_coeficientes (list of list of float): Matriz dos coeficientes.

    Retorna:
    ResultadoSassenfeld: Resultado do critério de Sassenfeld.
    """
    n = len(matriz_coeficientes)  # Obtém o número de linhas da matriz
    betas = [0] * n  # Inicializa a lista de betas

    # Verifica se os coeficientes na diagonal principal são positivos
    criterio_aplicavel = True
    for i in range(n):
        if matriz_coeficientes[i][i] <= 0:  # Se algum coeficiente diagonal é não positivo
            criterio_aplicavel = False
            break

    if not criterio_aplicavel:
        return ResultadoSassenfeld(False, False, 0)  # Retorna resultado se critério não for aplicável

    # Calcula os betas para cada linha
    for i in range(n):
        soma = 0.0
        for j in range(n):
            if j != i:
                soma += abs(matriz_coeficientes[i][j])
        betas[i] = soma / abs(matriz_coeficientes[i][i])

    # Encontra o valor máximo de beta
    max_beta = max(betas)

    # Verifica se os betas satisfazem o critério de Sassenfeld
    criterio_atendido = max_beta < 1.0

    return ResultadoSassenfeld(True, criterio_atendido, max_beta)  # Retorna o resultado do critério


if __name__ == "__main__":
    # Carregar o arquivo Excel
    print("Faça o upload da matriz A")  # Pede ao usuário para fazer o upload da matriz A
    uploaded = files.upload()  # Realiza o upload do arquivo

    # Obter o nome do arquivo
    file_name = list(uploaded.keys())[0]  # Obtém o nome do arquivo carregado

    # Ler a matriz A do arquivo Excel
    df_A = pd.read_excel(file_name, header=None)  # Lê o arquivo Excel em um DataFrame

    # Converter o dataframe para um array numpy
    matriz = df_A.to_numpy()  # Converte o DataFrame para um array numpy

    # Verificar a convergência usando os diferentes critérios
    print("Critério das Linhas:", criterio_linhas(matriz))  # Verifica e imprime o resultado do critério das linhas
    print("Critério das Colunas:", criterio_colunas(matriz))  # Verifica e imprime o resultado do critério das colunas
    resultado = criterio_sassenfeld(matriz)  # Verifica o critério de Sassenfeld
    if resultado.criterio_aplicavel:
        print("O critério de Sassenfeld é aplicável à matriz fornecida.")
        print("Critério de Sassenfeld atendido:", resultado.criterio_atendido)
        print("Máximo Beta encontrado:", resultado.max_beta)
    else:
        print("O critério de Sassenfeld não é aplicável à matriz fornecida.")
```

```python
import numpy as np  # Importa a biblioteca numpy para operações numéricas
import pandas as pd  # Importa a biblioteca pandas para manipulação de dados
from google.colab import files  # Importa o módulo files do Google Colab para upload de arquivos


# Implementação do método SOR
def metodo_SOR(A, b, omega, epsilon, max_iteracoes):
    """
    Método de Sobrerrelaxação Sucessiva (SOR) para resolver sistemas lineares.

    Parâmetros:
    A (ndarray): Matriz dos coeficientes.
    b (ndarray): Vetor dos termos independentes.
    omega (float): Parâmetro de relaxação.
    epsilon (float): Critério de parada (tolerância para o erro).
    max_iteracoes (int): Número máximo de iterações.

    Retorna:
    tuple: Vetor solução e número de iterações realizadas.
    """
    n = len(b)  # Número de elementos no vetor b
    x = np.zeros(n)  # Vetor inicial (pode ser qualquer valor inicial)
    x_anterior = np.zeros(n)  # Vetor da iteração anterior
    erro = float('inf')  # Inicializa o erro com infinito
    iteracoes = 0  # Contador de iterações

    while erro > epsilon and iteracoes < max_iteracoes:
        # Realiza uma iteração do método SOR
        for i in range(n):
            sigma = sum(A[i][j] * x[j] for j in range(n) if j != i)  # Soma dos produtos dos coeficientes A e das estimativas de x
            x[i] = (1 - omega) * x[i] + (omega / A[i][i]) * (b[i] - sigma)  # Atualiza x[i] com o valor calculado

        # Calcula o erro
        erro = np.linalg.norm(x - x_anterior, ord=np.inf)  # Calcula o erro como a norma infinita da diferença entre x e x_anterior
        x_anterior[:] = x  # Atualiza x_anterior com os valores de x
        iteracoes += 1  # Incrementa o contador de iterações

    return x, iteracoes  # Retorna o vetor solução e o número de iterações realizadas

# Método para encontrar o omega ótimo usando busca linear
def encontrar_WOtimo(A, b, epsilon, max_iteracoes):
    """
    Encontra o valor ótimo de omega para o método SOR usando busca linear.

    Parâmetros:
    A (ndarray): Matriz dos coeficientes.
    b (ndarray): Vetor dos termos independentes.
    epsilon (float): Critério de parada (tolerância para o erro).
    max_iteracoes (int): Número máximo de iterações.

    Retorna:
    float: Valor ótimo de omega.
    """
    omega_min = 1.0  # Valor mínimo de omega
    omega_max = 2.0  # Valor máximo de omega
    incremento = 0.01  # Incremento para busca linear
    num_iteracoes_min = float('inf')  # Inicializa o número mínimo de iterações com infinito
    WOtimo = omega_min  # Inicializa o valor ótimo de omega

    # Realiza a busca linear para encontrar o omega ótimo
    for omega in np.arange(omega_min, omega_max + incremento, incremento):
        _, num_iteracoes = metodo_SOR(A, b, omega, epsilon, max_iteracoes)  # Executa o método SOR com o valor atual de omega
        if num_iteracoes < num_iteracoes_min:
            num_iteracoes_min = num_iteracoes  # Atualiza o número mínimo de iterações
            WOtimo = omega  # Atualiza o valor ótimo de omega

    return WOtimo  # Retorna o valor ótimo de omega

# Método principal para teste
if __name__ == "__main__":
    # Carregar arquivos Excel
    print("Faça upload dos arquivos Excel.")  # Solicita o upload dos arquivos
    uploaded_A = files.upload()  # Realiza o upload do arquivo da matriz A
    uploaded_b = files.upload()  # Realiza o upload do arquivo do vetor b

    # Ler a matriz A e o vetor b dos arquivos Excel
    nome_arquivo_A = list(uploaded_A.keys())[0]  # Obtém o nome do arquivo da matriz A
    nome_arquivo_b = list(uploaded_b.keys())[0]  # Obtém o nome do arquivo do vetor b

    df_A = pd.read_excel(nome_arquivo_A, header=None)  # Lê o arquivo Excel da matriz A
    df_b = pd.read_excel(nome_arquivo_b, header=None)  # Lê o arquivo Excel do vetor b

    A = df_A.to_numpy()  # Converte o DataFrame da matriz A para um array numpy
    b = df_b.to_numpy().flatten()  # Converte o DataFrame do vetor b para um array numpy e achata para obter um vetor

    # Parâmetros do método SOR
    epsilon = 1e-6  # Critério de parada (tolerância para o erro)
    max_iteracoes = 1000  # Número máximo de iterações

    # Encontrar o omega ótimo
    WOtimo = encontrar_WOtimo(A, b, epsilon, max_iteracoes)  # Calcula o valor ótimo de omega
    print("WOtimo:", WOtimo)  # Imprime o valor ótimo de omega
```

```python
import sympy as sp

def newton_raphson(estimativa_inicial, funcao_expr, tolerancia, max_iteracoes=1000):
    # Define o símbolo 'x' para ser usado nas expressões simbólicas
    x = sp.symbols('x')
    # Converte a expressão simbólica da função em uma função numérica usando numpy
    funcao = sp.lambdify(x, funcao_expr, 'numpy')
    # Calcula a derivada simbólica da função
    derivada_expr = sp.diff(funcao_expr, x)
    # Converte a expressão simbólica da derivada em uma função numérica usando numpy
    derivada = sp.lambdify(x, derivada_expr, 'numpy')

    # Inicializa a estimativa atual com a estimativa inicial fornecida
    estimativa_atual = estimativa_inicial
    # Inicializa a estimativa anterior como None
    estimativa_anterior = None
    # Inicializa o contador de iterações
    iteracoes = 0

    # Loop até que a diferença entre estimativas seja menor que a tolerância ou que não haja estimativa anterior
    while estimativa_anterior is None or abs(estimativa_atual - estimativa_anterior) > tolerancia:
        # Verifica se o número máximo de iterações foi atingido
        if iteracoes >= max_iteracoes:
            raise Exception("Número máximo de iterações atingido")

        # Atualiza a estimativa anterior para a estimativa atual
        estimativa_anterior = estimativa_atual
        # Calcula o valor da função na estimativa anterior
        f = funcao(estimativa_anterior)
        # Calcula o valor da derivada na estimativa anterior
        f_prime = derivada(estimativa_anterior)

        # Verifica se a derivada é zero para evitar divisão por zero
        if f_prime == 0:
            raise Exception("Derivada zero encontrada, escolha uma estimativa inicial diferente")

        # Atualiza a estimativa atual usando a fórmula de Newton-Raphson
        estimativa_atual = estimativa_anterior - f / f_prime
        # Incrementa o contador de iterações
        iteracoes += 1

    # Retorna a estimativa atual, que é a raiz encontrada
    return estimativa_atual

if __name__ == "__main__":
    # Solicita ao usuário a função como uma string
    funcao_str = input("Digite a função f(x) (ex: 'x**3 - 2*x - 5'): ")

    # Converte a string da função em uma expressão simbólica do sympy
    x = sp.symbols('x')
    funcao_expr = sp.sympify(funcao_str)

    # Solicita ao usuário a estimativa inicial e converte para float
    estimativa_inicial = float(input("Digite a estimativa inicial: "))

    # Solicita ao usuário a tolerância e converte para float
    tolerancia = float(input("Digite a tolerância: "))

    try:
        # Chama a função newton_raphson para encontrar a raiz usando os parâmetros fornecidos
        raiz = newton_raphson(estimativa_inicial, funcao_expr, tolerancia)
        # Imprime a raiz encontrada
        print("A raiz encontrada é:", raiz)
    except Exception as e:
        # Captura e imprime qualquer exceção que ocorra durante a execução
        print("Erro:", e)
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from google.colab import files

# Função para dividir o dataset em treinamento e teste
def train_test_split(X, y, test_size=0.2, random_state=None):
    # Define uma semente para o gerador de números aleatórios, se fornecida
    if random_state is not None:
        np.random.seed(random_state)
    # Cria um array de índices baseado no número de amostras em X
    indices = np.arange(X.shape[0])
    # Embaralha os índices aleatoriamente
    np.random.shuffle(indices)
    # Calcula o ponto de divisão baseado no tamanho do teste
    split = int(X.shape[0] * (1 - test_size))
    # Seleciona os índices para os dados de treinamento e teste
    train_indices = indices[:split]
    test_indices = indices[split:]
    # Retorna os dados de treinamento e teste, tanto para X quanto para y
    return X[train_indices], X[test_indices], y[train_indices], y[test_indices]

# Função para normalizar os dados
def normalize(X, mean=None, std=None):
    # Calcula a média das colunas se não fornecida
    if mean is None:
        mean = np.mean(X, axis=0)
    # Calcula o desvio padrão das colunas se não fornecido
    if std is None:
        std = np.std(X, axis=0)
    # Normaliza os dados (X - média) / desvio padrão
    return (X - mean) / std, mean, std

# Função para calcular a distância Euclidiana
def euclidean_distance(a, b):
    # Calcula a raiz quadrada da soma dos quadrados das diferenças
    return np.sqrt(np.sum((a - b) ** 2))

# Função para calcular a distância Manhattan
def manhattan_distance(a, b):
    # Calcula a soma das diferenças absolutas
    return np.sum(np.abs(a - b))

# Função para calcular a distância Chebyshev
def chebyshev_distance(a, b):
    # Calcula a máxima diferença absoluta
    return np.max(np.abs(a - b))

# Função para encontrar os k vizinhos mais próximos
def get_neighbors(X_train, y_train, test_instance, k, distance_func):
    distances = []
    # Para cada ponto de treinamento
    for i in range(X_train.shape[0]):
        # Calcula a distância entre o ponto de teste e o ponto de treinamento
        dist = distance_func(test_instance, X_train[i])
        # Armazena a distância e o rótulo do ponto de treinamento
        distances.append((y_train[i], dist))
    # Ordena as distâncias em ordem crescente
    distances.sort(key=lambda x: x[1])
    # Seleciona os k vizinhos mais próximos
    neighbors = distances[:k]
    # Retorna os rótulos dos vizinhos
    return [neighbor[0] for neighbor in neighbors]

# Função para fazer a previsão usando KNN
def predict(X_train, y_train, X_test, k, distance_func):
    predictions = []
    # Para cada ponto de teste
    for test_instance in X_test:
        # Encontra os k vizinhos mais próximos
        neighbors = get_neighbors(X_train, y_train, test_instance, k, distance_func)
        # Seleciona o rótulo mais comum entre os vizinhos
        prediction = max(set(neighbors), key=neighbors.count)
        # Adiciona a previsão à lista de previsões
        predictions.append(prediction)
    # Converte a lista de previsões em um array numpy
    return np.array(predictions)

# Função para calcular a acurácia
def accuracy_score(y_true, y_pred):
    # Calcula a proporção de previsões corretas
    return np.sum(y_true == y_pred) / len(y_true)

if __name__ == "__main__":
    # Pedir ao usuário para fazer upload do arquivo da matriz A
    print("Faça upload do arquivo da matriz A:")
    uploaded_A = files.upload()

    # Obter o nome do arquivo de upload
    file_name_A = list(uploaded_A.keys())[0]

    # Ler a matriz A do arquivo Excel
    df_A = pd.read_excel(file_name_A, header=None)

    # Verificar e tratar valores ausentes
    if df_A.isnull().values.any():
        print("Existem valores ausentes no dataset. Preenchendo com a média de cada coluna.")
        df_A = df_A.fillna(df_A.mean())

    # Converter o dataframe para arrays numpy
    A = df_A.to_numpy()

    # Separando features (X) e rótulos (y)
    X = A[:, :-1].astype(float)  # Todas as colunas menos a última são features
    y = A[:, -1]  # A última coluna é o rótulo

    # Convertendo rótulos (y) de notas para números
    label_encoder = LabelEncoder()
    y = label_encoder.fit_transform(y)

    # Dividindo os dados em conjunto de treinamento e teste
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

    # Normalizando os dados de treinamento
    X_train, mean, std = normalize(X_train)

    # Normalizando os dados de teste com os mesmos parâmetros do treinamento
    X_test, _, _ = normalize(X_test, mean, std)

    # Verificando os dados de treinamento e teste
    print("Dados de treinamento (X_train):\n", X_train)
    print("Dados de teste (X_test):\n", X_test)
    print("Rótulos de treinamento (y_train):\n", y_train)
    print("Rótulos de teste (y_test):\n", y_test)

    # Função para testar diferentes normas
    def test_knn(norm_name, distance_func):
        k = 1  # Ajustando o valor de k
        y_pred = predict(X_train, y_train, X_test, k, distance_func)
        accuracy = accuracy_score(y_test, y_pred) * 100  # Converter para porcentagem
        print(f'Acurácia com norma {norm_name}: {accuracy:.2f}%')

        # Mostrar os valores previstos e reais
        for true, pred in zip(y_test, y_pred):
            true_label = label_encoder.inverse_transform([true])[0]
            pred_label = label_encoder.inverse_transform([pred])[0]
            print(f'Verdadeiro: {true_label}, Previsto: {pred_label}')

    # Testando diferentes normas
    test_knn('euclidean', euclidean_distance)
    test_knn('manhattan', manhattan_distance)
    test_knn('chebyshev', chebyshev_distance)
```

```python
import numpy as np
from sympy import symbols, Matrix, det
import pandas as pd  # Importa a biblioteca pandas para manipulação de dados
from google.colab import files  # Importa o módulo files do Google Colab para upload de arquivos

def polinomio_caracteristico(A):
    # Verifique se a matriz é quadrada
    if A.shape[0] != A.shape[1]:  # Verifica se o número de linhas é diferente do número de colunas
        raise ValueError("A matriz deve ser quadrada")  # Levanta um erro se a matriz não for quadrada

    # Obtenha a ordem da matriz
    n = A.shape[0]  # Obtém o número de linhas da matriz (que deve ser igual ao número de colunas)

    # Defina a variável simbólica λ
    λ = symbols('λ')  # Cria o símbolo lambda (variável simbólica)

    # Construa a matriz simbólica A - λI
    I = np.eye(n)  # Cria uma matriz identidade do mesmo tamanho que A
    A_lambda_I = Matrix(A) - λ * Matrix(I)  # Calcula A - λI, onde λ é multiplicado elemento por elemento

    # Calcule o determinante
    polinomio = det(A_lambda_I)  # Calcula o determinante da matriz simbólica, que é o polinômio característico

    return polinomio

# Exemplo de uso
if __name__ == "__main__":
    # Carregar o arquivo Excel
    print("Faça upload da matriz A")
    uploaded = files.upload()  # Pede ao usuário para fazer upload do arquivo

    # Obter o nome do arquivo
    file_name = list(uploaded.keys())[0]  # Obtém o nome do arquivo carregado

    # Ler a matriz A do arquivo Excel
    df_A = pd.read_excel(file_name, header=None)  # Lê o arquivo Excel e armazena em um DataFrame

    # Converter o dataframe para um array numpy
    A = df_A.to_numpy()  # Converte o DataFrame para um array numpy

    print("A matriz A é:")
    print(A)

    # Calcula o polinômio característico da matriz A
    polinomio = polinomio_caracteristico(A)
    print("\nO polinômio característico de A é:")
    print(polinomio)
```

```python
import numpy as np
import pandas as pd  # Importa a biblioteca pandas para manipulação de dados
from google.colab import files  # Importa o módulo files do Google Colab para upload de arquivos

def svd_manual(A):
    # Passo 1: Calcular A * A^T e A^T * A
    AAT = np.dot(A, A.T)
    ATA = np.dot(A.T, A)

    # Passo 2: Calcular os autovalores e autovetores de A * A^T (para U)
    autovalores_AAT, autovetores_U = np.linalg.eig(AAT)

    # Passo 3: Calcular os autovalores e autovetores de A^T * A (para V)
    autovalores_ATA, autovetores_V = np.linalg.eig(ATA)

    # Passo 4: Ordenar os autovalores e autovetores em ordem decrescente
    indices_ordenados_AAT = np.argsort(autovalores_AAT)[::-1]
    autovalores_ordenados_AAT = autovalores_AAT[indices_ordenados_AAT]
    U = autovetores_U[:, indices_ordenados_AAT]

    indices_ordenados_ATA = np.argsort(autovalores_ATA)[::-1]
    autovalores_ordenados_ATA = autovalores_ATA[indices_ordenados_ATA]
    V = autovetores_V[:, indices_ordenados_ATA]

    # Passo 5: Calcular os valores singulares
    valores_singulares = np.sqrt(autovalores_ordenados_AAT)

    # Passo 6: Construir a matriz Sigma como uma matriz diagonal
    Sigma = np.zeros((A.shape[0], A.shape[1]))
    np.fill_diagonal(Sigma, valores_singulares)

    # Passo 7: Calcular U e V
    U = autovetores_U[:, indices_ordenados_AAT]
    V = autovetores_V[:, indices_ordenados_ATA]

    # Passo 8: Normalizar U e V
    U = np.dot(A, np.dot(V, np.linalg.pinv(Sigma)))  # Corrigindo U baseado em A, V e Sigma
    V = V / np.linalg.norm(V, axis=0)

    # Transpor V para obter V^T
    VT = V.T

    # Verificar a reconstrução da matriz original
    A_reconstructed = np.dot(U, np.dot(Sigma, VT))
    print("\nMatriz A reconstruída:")
    print(A_reconstructed)

    return U, Sigma, VT

if __name__ == "__main__":
  # Pedir ao usuário para fazer upload do arquivo da matriz A
  print("Faça upload do arquivo da matriz A:")
  uploaded_A = files.upload()

  # Obter os nomes dos arquivos de upload
  file_name_A = list(uploaded_A.keys())[0]

  # Ler a matriz A e o vetor b dos arquivos Excel
  df_A = pd.read_excel(file_name_A, header=None)

  # Converter os dataframes para arrays numpy
  A = df_A.to_numpy()

  U, Sigma, VT = svd_manual(A)
  print("Matriz U:")
  print(U)
  print("\nMatriz Sigma:")
  print(Sigma)
  print("\nMatriz VT:")
  print(VT)
```

Letra D(Gram-Schmidt)

```python
import numpy as np
import pandas as pd  # Importa a biblioteca pandas para manipulação de dados
from google.colab import files  # Importa o módulo files do Google Colab para upload de arquivos

def gram_schmidt(matriz):
    # Verificar se a matriz é válida
    if not isinstance(matriz, np.ndarray):
        raise ValueError("A entrada deve ser uma matriz numpy.")

    if matriz.ndim != 2:
        raise ValueError("A matriz de entrada deve ser bidimensional.")

    # Verificar se os vetores são linearmente independentes
    if np.linalg.matrix_rank(matriz) < matriz.shape[1]:
        raise ValueError("Os vetores fornecidos não são linearmente independentes.")

    # Obter o número de vetores (colunas) na matriz
    n = matriz.shape[1]

    # Inicializar a matriz para os vetores ortogonais
    Q = np.zeros_like(matriz, dtype=float)

    # Iterar sobre cada vetor na matriz
    for i in range(n):
        # Começar com o vetor atual
        u = matriz[:, i]

        # Subtrair as projeções de u em cada um dos vetores q anteriores
        for j in range(i):
            u -= np.dot(Q[:, j], matriz[:, i]) * Q[:, j]

        # Verificar se a norma é zero (o que significaria que os vetores são linearmente dependentes)
        norma_u = np.linalg.norm(u)
        if norma_u == 0:
            raise ValueError(f"O vetor {i} é uma combinação linear dos vetores anteriores, indicando dependência linear.")

        # Normalizar o vetor
        Q[:, i] = u / norma_u

    return Q

if __name__ == "__main__":
  try:
    # Pedir ao usuário para fazer upload do arquivo da matriz A
    print("Faça upload do arquivo da matriz A:")
    uploaded_A = files.upload()

    # Obter os nomes dos arquivos de upload
    file_name_A = list(uploaded_A.keys())[0]

    # Ler a matriz A e o vetor b dos arquivos Excel
    df_A = pd.read_excel(file_name_A, header=None)

    # Converter os dataframes para arrays numpy
    matriz = df_A.to_numpy(dtype=float)

    Q = gram_schmidt(matriz)
    print("Base ortonormal Q:")
    print(Q)

    # Verificar a ortogonalidade
    print("Q^T * Q (deve ser a matriz identidade):")
    print(np.dot(Q.T, Q))
  except ValueError as e:
    print(e)
```

Letra E(Problema de Leontief)

```python
import numpy as np
import pandas as pd  # Importa a biblioteca pandas para manipulação de dados
from google.colab import files  # Importa o módulo files do Google Colab para upload de arquivos

def resolver_problema_de_leontief(matriz_A, vetor_b):
    """
    Resolve o problema de Leontief dado uma matriz de insumo-produto A e um vetor de demanda final b.

    Parâmetros:
    matriz_A (numpy.ndarray): A matriz de insumo-produto (input-output matrix).
    vetor_b (numpy.ndarray): O vetor de demanda final.

    Retorna:
    numpy.ndarray: O vetor de produção que satisfaz a demanda final.
    """
    # Verificando se a matriz_A é quadrada
    if matriz_A.shape[0] != matriz_A.shape[1]:
        raise ValueError("A matriz de insumo-produto A deve ser quadrada.")

    # Verificando se as dimensões do vetor_b são compatíveis com a matriz_A
    if matriz_A.shape[0] != vetor_b.shape[0]:
        raise ValueError("O vetor de demanda final b deve ter o mesmo número de linhas que a matriz A.")

    # Criando a matriz identidade I
    matriz_identidade = np.identity(matriz_A.shape[0])

    # Calculando a matriz (I - A)
    matriz_I_menos_A = matriz_identidade - matriz_A

    # Calculando a inversa de (I - A)
    try:
        matriz_inversa = np.linalg.inv(matriz_I_menos_A)
    except np.linalg.LinAlgError:
        raise ValueError("A matriz (I - A) não é invertível.")

    # Calculando o vetor de produção c = ((I - A)^(-1)) * b
    vetor_c = np.dot(matriz_inversa, vetor_b)

    return vetor_c

# Exemplo de uso
if __name__ == "__main__":
  # Pedir ao usuário para fazer upload do arquivo da matriz A
  print("Faça upload do arquivo da matriz A:")
  uploaded_A = files.upload()

  # Pedir ao usuário para fazer upload do arquivo do vetor b
  print("Faça upload do arquivo do vetor b:")
  uploaded_b = files.upload()

  # Obter os nomes dos arquivos de upload
  file_name_A = list(uploaded_A.keys())[0]
  file_name_b = list(uploaded_b.keys())[0]

  # Ler a matriz A e o vetor b dos arquivos Excel
  df_A = pd.read_excel(file_name_A, header=None)
  df_b = pd.read_excel(file_name_b, header=None)

  # Converter os dataframes para arrays numpy
  matriz_A = df_A.to_numpy()
  vetor_b = df_b.to_numpy().flatten()  # Flatten para converter b em um vetor unidimensional

  # Resolvendo o problema de Leontief
  vetor_c = resolver_problema_de_leontief(matriz_A, vetor_b)

  # Exibindo o resultado
  print("Vetor de Produção (c):")
  print(vetor_c)
```

Letra F (Método da potência)

```python
import numpy as np
import pandas as pd
from google.colab import files  # Importa o módulo files do Google Colab para upload de arquivos

def norma(vetor):
    """
    Calcula a norma euclidiana de um vetor.
    """
    return np.sqrt(np.sum(vetor ** 2))

def metodo_da_potencia(matriz, vetor_inicial, max_iteracoes=1000, tolerancia=1e-10):
    """
    Aplica o método da potência para encontrar o autovalor dominante e o autovetor correspondente.

    Parâmetros:
    matriz (np.ndarray): A matriz quadrada para a qual queremos encontrar o autovalor e autovetor dominante.
    vetor_inicial (np.ndarray): Um vetor inicial para iniciar as iterações.
    max_iteracoes (int): O número máximo de iterações a serem realizadas. Padrão é 1000.
    tolerancia (float): A tolerância para a convergência do autovetor. Padrão é 1e-10.

    Retorna:
    autovalor (float): O autovalor dominante encontrado.
    autovetor (np.ndarray): O autovetor correspondente ao autovalor dominante.
    """
    vetor = vetor_inicial / norma(vetor_inicial)  # Normaliza o vetor inicial

    for iteracao in range(max_iteracoes):
        novo_vetor = np.dot(matriz, vetor)  # Multiplica a matriz pelo vetor
        novo_vetor = novo_vetor / norma(novo_vetor)  # Normaliza o novo vetor

        # Verifica a convergência
        if np.linalg.norm(novo_vetor - vetor) < tolerancia:
            break

        vetor = novo_vetor  # Atualiza o vetor para a próxima iteração

    # Calcula o autovalor dominante
    autovalor = np.dot(vetor.T, np.dot(matriz, vetor)) / np.dot(vetor.T, vetor)

    return autovalor, vetor

def precondicionar_matriz(matriz, sigma):
    """
    Aplica pré-condicionamento à matriz subtraindo sigma vezes a identidade.

    Parâmetros:
    matriz (np.ndarray): A matriz quadrada original.
    sigma (float): O valor a ser subtraído da diagonal da matriz.

    Retorna:
    np.ndarray: A matriz pré-condicionada.
    """
    identidade = np.eye(matriz.shape[0])
    return matriz - sigma * identidade

# Exemplo de uso
if __name__ == "__main__":
  # Pedir ao usuário para fazer upload do arquivo da matriz A
  print("Faça upload do arquivo da matriz A:")
  uploaded_A = files.upload()

  # Pedir ao usuário para fazer upload do arquivo do vetor b
  print("Faça upload do arquivo do vetor b:")
  uploaded_b = files.upload()

  # Obter os nomes dos arquivos de upload
  file_name_A = list(uploaded_A.keys())[0]
  file_name_b = list(uploaded_b.keys())[0]

  # Ler a matriz A e o vetor b dos arquivos Excel
  df_A = pd.read_excel(file_name_A, header=None)
  df_b = pd.read_excel(file_name_b, header=None)

  # Converter os dataframes para arrays numpy
  A = df_A.to_numpy()
  vetor_inicial = df_b.to_numpy().flatten()  # Flatten para converter b em um vetor unidimensional

  # Resultados sem pré-condicionamento
  autovalor, autovetor = metodo_da_potencia(A, vetor_inicial)
  print("Sem pré-condicionamento:")
  print("Autovalor dominante:", autovalor)
  print("Autovetor correspondente:", autovetor)

  # Valor de sigma para pré-condicionamento (estimativa do autovalor dominante)
  sigma = autovalor

  # Matriz pré-condicionada
  A_precondicionada = precondicionar_matriz(A, sigma)

  # Resultados com pré-condicionamento
  autovalor_precond, autovetor_precond = metodo_da_potencia(A_precondicionada, vetor_inicial)
  # Reajustar o autovalor encontrado após pré-condicionamento
  autovalor_precond += sigma

  print("\nCom pré-condicionamento:")
  print("Autovalor dominante:", autovalor_precond)
  print("Autovetor correspondente:", autovetor_precond)
```

Letra G, H e J(Método da descida)

```python
import numpy as np
import pandas as pd  # Importa a biblioteca pandas para manipulação de dados
from google.colab import files  # Importa o módulo files do Google Colab para upload de arquivos

# Função para verificar se a matriz é simétrica
def is_symmetric(A, tol=1e-10):
    return np.allclose(A, A.T, atol=tol)

# Função para verificar se a matriz é positiva definida
def is_positive_definite(A):
    try:
        np.linalg.cholesky(A)
        return True
    except np.linalg.LinAlgError:
        return False

# Algoritmo de Descida de Gradiente
def gradient_descent(A, b, tol=1e-10, max_iter=1000):
    x = np.zeros_like(b)  # Inicializa o vetor solução com zeros
    r = b - np.dot(A, x)  # Calcula o resíduo inicial
    iter_count = 0  # Inicializa o contador de iterações

    # Loop até a norma do resíduo ser menor que a tolerância ou atingir o número máximo de iterações
    while np.linalg.norm(r) > tol and iter_count < max_iter:
        alpha = np.dot(r, r) / np.dot(r, np.dot(A, r))  # Calcula o passo (tamanho do passo)
        x = x + alpha * r  # Atualiza a solução
        r = b - np.dot(A, x)  # Atualiza o resíduo
        iter_count += 1  # Incrementa o contador de iterações

    return x, iter_count  # Retorna a solução e o número de iterações

# Algoritmo de Descida de Gradiente Precondicionado
def preconditioned_gradient_descent(A, b, M, tol=1e-10, max_iter=1000):
    x = np.zeros_like(b)  # Inicializa o vetor solução com zeros
    r = b - np.dot(A, x)  # Calcula o resíduo inicial
    z = np.linalg.solve(M, r)  # Resolve Mz = r para z
    iter_count = 0  # Inicializa o contador de iterações

    # Loop até a norma do resíduo ser menor que a tolerância ou atingir o número máximo de iterações
    while np.linalg.norm(r) > tol and iter_count < max_iter:
        alpha = np.dot(r, z) / np.dot(z, np.dot(A, z))  # Calcula o passo (tamanho do passo)
        x = x + alpha * z  # Atualiza a solução
        r = b - np.dot(A, x)  # Atualiza o resíduo
        z = np.linalg.solve(M, r)  # Resolve Mz = r para z
        iter_count += 1  # Incrementa o contador de iterações

    return x, iter_count  # Retorna a solução e o número de iterações

# Código principal
if __name__ == "__main__":
    # Pedir ao usuário para fazer upload do arquivo da matriz A
    print("Faça upload do arquivo da matriz A:")
    uploaded_A = files.upload()

    # Pedir ao usuário para fazer upload do arquivo do vetor b
    print("Faça upload do arquivo do vetor b:")
    uploaded_b = files.upload()

    # Obter os nomes dos arquivos de upload
    file_name_A = list(uploaded_A.keys())[0]
    file_name_b = list(uploaded_b.keys())[0]

    # Ler a matriz A e o vetor b dos arquivos Excel
    df_A = pd.read_excel(file_name_A, header=None)
    df_b = pd.read_excel(file_name_b, header=None)

    # Converter os dataframes para arrays numpy
    A = df_A.to_numpy(dtype=float)
    b = df_b.to_numpy(dtype=float).flatten()  # Flatten para converter b em um vetor unidimensional

    # Verificando se a matriz A é simétrica e positiva definida
    if not is_symmetric(A) or not is_positive_definite(A):
        if not is_positive_definite(A):
            print("A matriz A não é positiva definida")
        else:
            print("A matriz A não é simétrica")
    else:
        # Matriz de Jacobi como precondicionador
        M = np.diag(np.diag(A))

        # Resolvendo sem precondicionamento
        x_gd, iter_gd = gradient_descent(A, b)

        # Resolvendo com precondicionamento
        x_pgd, iter_pgd = preconditioned_gradient_descent(A, b, M)

        print("Resultados sem precondicionamento:")
        print("Solução x:", x_gd)
        print("Número de iterações:", iter_gd)

        print("\nResultados com precondicionamento:")
        print("Solução x:", x_pgd)
        print("Número de iterações:", iter_pgd)
```

Letra I(Gradiente conjugado)

```python
import numpy as np
import pandas as pd  # Importa a biblioteca pandas para manipulação de dados
from google.colab import files  # Importa o módulo files do Google Colab para upload de arquivos

def is_symmetric(A, tol=1e-10):
    return np.allclose(A, A.T, atol=tol)

def is_positive_definite(A):
    try:
        np.linalg.cholesky(A)
        return True
    except np.linalg.LinAlgError:
        return False

def gradiente_conjugado(A, b, tol=1e-10, max_iter=1000):
    # Inicializa a solução x com zeros
    x = np.zeros_like(b)
    # Calcula o resíduo inicial r = b - Ax
    r = b - np.dot(A, x)
    # Inicializa a direção de busca p com o resíduo inicial
    p = r.copy()
    # Calcula o produto interno inicial do resíduo
    rsold = np.dot(r, r)
    iter_count = 0  # Contador de iterações

    while np.sqrt(rsold) > tol and iter_count < max_iter:
        # Calcula Ap, onde A é a matriz do sistema e p é a direção de busca atual
        Ap = np.dot(A, p)
        # Calcula o fator de passo alpha
        alpha = rsold / np.dot(p, Ap)
        # Atualiza a solução x
        x = x + alpha * p
        # Atualiza o resíduo
        r = r - alpha * Ap
        # Calcula o novo produto interno do resíduo
        rsnew = np.dot(r, r)

        # Verifica a condição de parada baseada na tolerância
        if np.sqrt(rsnew) < tol:
            break

        # Atualiza a direção de busca p
        p = r + (rsnew / rsold) * p
        # Atualiza rsold para a próxima iteração
        rsold = rsnew
        iter_count += 1  # Incrementa o contador de iterações

    return x, iter_count

def gradiente_conjugado_precondicionado(A, b, M, tol=1e-10, max_iter=1000):
    # Inicializa a solução x com zeros
    x = np.zeros_like(b)
    # Calcula o resíduo inicial r = b - Ax
    r = b - np.dot(A, x)
    # Resolve Mz = r para obter z
    z = np.linalg.solve(M, r)
    # Inicializa a direção de busca p com z
    p = z.copy()
    # Calcula o produto interno inicial do resíduo com o precondicionamento
    rsold = np.dot(r, z)
    iter_count = 0  # Contador de iterações

    while np.sqrt(rsold) > tol and iter_count < max_iter:
        # Calcula Ap, onde A é a matriz do sistema e p é a direção de busca atual
        Ap = np.dot(A, p)
        # Calcula o fator de passo alpha
        alpha = rsold / np.dot(p, Ap)
        # Atualiza a solução x
        x = x + alpha * p
        # Atualiza o resíduo
        r = r - alpha * Ap
        # Resolve Mz = r para obter o novo z
        z = np.linalg.solve(M, r)
        # Calcula o novo produto interno do resíduo com o precondicionamento
        rsnew = np.dot(r, z)

        # Verifica a condição de parada baseada na tolerância
        if np.sqrt(rsnew) < tol:
            break

        # Atualiza a direção de busca p
        p = z + (rsnew / rsold) * p
        # Atualiza rsold para a próxima iteração
        rsold = rsnew
        iter_count += 1  # Incrementa o contador de iterações

    return x, iter_count

if __name__ == "__main__":
  # Pedir ao usuário para fazer upload do arquivo da matriz A
  print("Faça upload do arquivo da matriz A:")
  uploaded_A = files.upload()

  # Pedir ao usuário para fazer upload do arquivo do vetor b
  print("Faça upload do arquivo do vetor b:")
  uploaded_b = files.upload()

  # Obter os nomes dos arquivos de upload
  file_name_A = list(uploaded_A.keys())[0]
  file_name_b = list(uploaded_b.keys())[0]

  # Ler a matriz A e o vetor b dos arquivos Excel
  df_A = pd.read_excel(file_name_A, header=None)
  df_b = pd.read_excel(file_name_b, header=None)

  # Converter os dataframes para arrays numpy
  A = df_A.to_numpy(dtype=float)
  b = df_b.to_numpy(dtype=float).flatten()  # Flatten para converter b em um vetor unidimensional

  if not is_symmetric(A) or not is_positive_definite(A):
    if not is_positive_definite(A):
      print("A matriz A não é positiva definida")
    else:
      print("A matriz A não é simétrica")


  else:
    # Matriz de Jacobi como precondicionador, que é a diagonal de A
    M = np.diag(np.diag(A))

    # Resolvendo sem precondicionamento
    x_gc, iter_gc = gradiente_conjugado(A, b)

    # Resolvendo com precondicionamento
    x_gcp, iter_gcp = gradiente_conjugado_precondicionado(A, b, M)

    # Impressão dos resultados
    print("Resultados sem precondicionamento:")
    print("Solução x:", x_gc)
    print("Número de iterações:", iter_gc)

    print("\nResultados com precondicionamento:")
    print("Solução x:", x_gcp)
    print("Número de iterações:", iter_gcp)
```
